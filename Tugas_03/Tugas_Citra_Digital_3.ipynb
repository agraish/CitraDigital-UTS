```python
from google.colab import files
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

uploaded = files.upload()
```

```python
nama_file = "Ijazah 1.jpeg"
img = Image.open(nama_file)

plt.figure(figsize=(8, 5))
plt.imshow(img)
plt.title("Ijazah 1 - Citra Asli (RGB)")
plt.axis("off")
plt.show()
```

```python
gray = img.convert("L")

plt.figure(figsize=(8, 5))
plt.imshow(gray, cmap="gray")
plt.title("Ijazah 1 - Grayscale")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(12, 5))

# Histogram citra RGB
plt.subplot(1, 2, 1)
plt.hist(np.array(img).ravel(), bins=256, range=(0, 255))
plt.title("Histogram Ijazah 1 - RGB")
plt.xlabel("Intensitas Piksel")
plt.ylabel("Jumlah Piksel")
```

```python
# Histogram grayscale
plt.subplot(1, 2, 2)
plt.hist(np.array(gray).ravel(), bins=256, range=(0, 255))
plt.title("Histogram Ijazah 1 - Grayscale")
plt.xlabel("Intensitas Piksel")
plt.ylabel("Jumlah Piksel")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))

# Histogram citra RGB
plt.subplot(1, 2, 1)
plt.hist(np.array(img).ravel(), bins=256, range=(0, 255))
plt.title("Histogram Ijazah 1 - RGB")
plt.xlabel("Intensitas Piksel")
plt.ylabel("Jumlah Piksel")

# Histogram grayscale
plt.subplot(1, 2, 2)
plt.hist(np.array(gray).ravel(), bins=256, range=(0, 255))
plt.title("Histogram Ijazah 1 - Grayscale")
plt.xlabel("Intensitas Piksel")
plt.ylabel("Jumlah Piksel")

plt.tight_layout()
plt.show()
```

```python
gray_array = np.array(gray)

print("Mean       :", gray_array.mean())
print("Std Dev    :", gray_array.std())
print("Minimum    :", gray_array.min())
print("Maximum    :", gray_array.max())
```

```python
from PIL import ImageOps

enhanced = ImageOps.autocontrast(gray)

plt.figure(figsize=(8, 5))
plt.imshow(enhanced, cmap="gray")
plt.title("Ijazah 1 - Setelah Contrast Stretching")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(gray_array.ravel(), bins=256, range=(0, 255))
plt.title("Sebelum Enhancement")
plt.xlabel("Intensitas Piksel")
plt.ylabel("Jumlah Piksel")

plt.subplot(1, 2, 2)
enhanced_array = np.array(enhanced)
plt.hist(enhanced_array.ravel(), bins=256, range=(0, 255))
plt.title("Sesudah Contrast Stretching")
plt.xlabel("Intensitas Piksel")
plt.ylabel("Jumlah Piksel")

plt.tight_layout()
plt.show()
```
